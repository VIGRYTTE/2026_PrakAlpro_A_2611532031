## 2.1 Dasar Teori
Pengenalan tentang operator dalam python


1. Operator Aritmatika<br>
Operator aritmatika adalah operator yang digunakan untuk melakukan operasi matematika. Operator ini digunakan dalam berbagai proses perhitungan, seperti penjumlahan, pengurangan, perkalian, dan pembagian.

2. Operator Assignment<br>
Operator assignment atau operator penugasan digunakan untuk memberikan suatu nilai kepada variabel. Operator ini juga dapat digunakan untuk mengubah nilai yang sudah tersimpan pada sebuah variabel.

3. Operator Bitwise<br>
Operator bitwise merupakan operator yang digunakan untuk melakukan operasi pada tingkat bit dari suatu bilangan. Bilangan akan diproses berdasarkan bentuk binernya, yaitu menggunakan angka 0 dan 1.

4. Operator Keanggotaan<br>
Operator keanggotaan atau membership operator digunakan untuk memeriksa apakah suatu nilai terdapat di dalam sebuah kumpulan data.

5. Operator Identitas<br>
Operator identitas atau identity operator digunakan untuk memeriksa apakah dua variabel mengacu pada objek yang sama.

6. Operator Perbandingan<br>
Operator perbandingan digunakan untuk membandingkan dua nilai atau lebih. Hasil dari perbandingan biasanya berupa nilai True atau False, yang menunjukkan apakah suatu kondisi benar atau salah.

7. Operator Logika<br>
Operator logika digunakan untuk menggabungkan atau memanipulasi beberapa kondisi. Operator ini sering digunakan dalam pengambilan keputusan pada suatu program.

## 2.2 Kode Program

### 2.2.1 Program "Operator Aritmatika_2611532031"
</div>

```python

# buat program untuk operator aritmatika dalam python
# nama variabel ditambah 4 digit nim terakhir contoh: angka1_1234
# program ini menggunakan fungsi input()
# nilai yang di masukkan akan di konversi menjadi tipe data integer

angka1_2031 = int(input("input angka-1:"))
angka2_2031 = int(input("input angka-2:"))

# PENJUMLAHAN
hasil_2031 = angka1_2031 + angka2_2031
print("\nOperator Penjumlahan")
print("Hasil =", hasil_2031)

# PENGURANGAN
hasil_2031 = angka1_2031 - angka2_2031
print("\nOperator Pengurangan")
print("Hasil =", hasil_2031)

# PERKALIAN
hasil_2031 = angka1_2031 * angka2_2031
print("\nOperator Perkalian")
print("Hasil =", hasil_2031)

# PEMBAGIAN
if angka2_2031 != 0:
    hasil_2031 = angka1_2031 / angka2_2031
    print("\nOperator Pembagian")
    print("Hasil =", hasil_2031)

    hasil_2031 = angka1_2031 // angka2_2031
    print("\nOperator Pembagian Bulat")
    print("Hasil =", hasil_2031)

    hasil_2031 = angka1_2031 % angka2_2031
    print("\nOperator Sisa Bagi")
    print("Hasil =", hasil_2031)
else:
    print("angka kedua tidak boleh bernilai 0.")

# PANGKAT
hasil_2031 = angka1_2031 ** angka2_2031
print("\n Operator Pangkat")
print("Hasil=", hasil_2031)
```

<div style="font-family:'Times New Roman', Times, serif; font-size: 18px;">

Penjelasan:
1. Penjumlahan<br>
Angka pertama yaitu 7 dijumlahkan dengan angka kedua yaitu 8. Hasilnya adalah 15.<br>
7 + 8 = 15
2. Pengurangan<br>
Angka pertama dikurangi dengan angka kedua. Karena angka pertama lebih kecil dari angka kedua, hasilnya menjadi -1.
7 - 8 = -1
3. Perkalian<br>
Angka 7 dikalikan dengan angka 8 sehingga menghasilkan 56.<br>
7 × 8 = 56
4. Pembagian<br>
Angka 7 dibagi dengan angka 8. Hasilnya berupa bilangan desimal, yaitu 0.875.<br>
7 ÷ 8 = 0.875
5. Pembagian Bulat<br>
Pada pembagian bulat, hasil pembagian 7 dengan 8 hanya mengambil bagian bilangan bulatnya. Karena hasil pembagian adalah 0.875, maka hasil pembagian bulatnya adalah 0.<br>
7 // 8 = 0
6. Sisa Bagi<br>
Operasi sisa bagi digunakan untuk mengetahui sisa dari pembagian 7 dengan 8. Karena 7 masih lebih kecil dari 8, maka seluruh angka 7 menjadi sisa pembagian.<br>
7 % 8 = 7<br>
Jadi, hasilnya adalah 7.
7. Pangkat<br>
Pada operasi pangkat, angka pertama yaitu 7 menjadi bilangan dasar dan angka kedua yaitu 8 menjadi pangkat.
7⁸ = 5.764.801<br>
Sehingga hasil yang ditampilkan adalah 5764801.

### 2.2.2 Program "Operator Assignment_2611532031"
</div>

```python
# nama variabel ditambah 4 digit nim terakhir contoh: angka1_1234
# program ini menggunakan fungsi input()
# nilai yang dimasukkan akan dikonversi menjadi tipe data integer
# program operator assignment dalam python

angka1_2031 = int(input("input angka-1: "))
angka2_2031 = int(input("input angka-1: "))

print("\n nilai awal angka1 =", angka1_2031)
print("nilai angka2 =", angka2_2031)

# assignment biasa 
hasil_2031 = angka1_2031
print("\n assingmnet biasa (=)")
print("Hasil =", hasil_2031)

# assignment penambahan
hasil_2031 = angka1_2031
hasil_2031 += angka2_2031
print("\n assignment penambahan (+=)")
print("Hasil =", hasil_2031)

# assignment perkalian
hasil_2031 = angka1_2031
hasil_2031 *= angka2_2031
print("\n assignment perkalian (*=)")
print("Hasil =", hasil_2031)

# assignment pembagian, pembagian bulat, dan sisa bagi 
if angka2_2031 != 0:
    hasil_2031 = angka1_2031
    hasil_2031 /= angka2_2031
    print("\n assignment pembagian (/=)")
    print("Hasil =", hasil_2031)

    #operator tambahan
    hasil_2031 = angka1_2031
    hasil_2031 //= angka2_2031
    print("\n assignment pembagian bulat (//=)")
    print("Hasil =", hasil_2031)

    hasil_2031 = angka1_2031
    hasil_2031 %= angka2_2031
    print("\n assignment sisa bagi (%=)")
    print("Hasil =", hasil_2031)

else:
    print("\n pembagian tidak dapat dilakukan")
    print("angka kedua tidak boleh 0")
```

<div style="font-family:'Times New Roman', Times, serif; font-size: 18px;">
Penjelasan:

1. Nilai Awal<br>
Program menampilkan nilai awal dari kedua variabel. angka1 memiliki nilai 8, sedangkan angka2 memiliki nilai 5.<br>
2. Assignment Biasa (`=`)<br>
Assignment biasa digunakan untuk memberikan nilai dari angka1 kepada variabel hasil. Karena nilai angka1 adalah 8, maka hasil yang ditampilkan adalah 8.<br>
Hasil = 8
3. Assignment Penambahan (`+=`)<br>
Operator += digunakan untuk menambahkan nilai angka2 ke nilai yang sudah terdapat pada hasil.
8 + 5 = 13<br>
Jadi, hasil yang ditampilkan adalah 13.
4. Assignment Perkalian (`*=`)<br>
Operator *= digunakan untuk mengalikan nilai hasil dengan nilai angka2.
8 × 5 = 40<br>
Jadi, hasil yang ditampilkan adalah 40.
5. Assignment Pembagian (`/=`)<br>
Operator /= digunakan untuk membagi nilai hasil dengan angka2.
8 ÷ 5 = 1.6<br>
Karena pembagian biasa dapat menghasilkan bilangan desimal, maka output yang diperoleh adalah 1.6.
6. Assignment Pembagian Bulat (`//=`)<br>
Operator //= digunakan untuk melakukan pembagian dan mengambil hasil bilangan bulatnya.

Bagian bilangan bulat dari 1.6 adalah 1, sehingga hasil yang ditampilkan adalah 1.

### 2.2.3 Program "Operator Bitwise_2611532031"

```python
# nama variabel ditambah 4 digit nim terakhir contoh: angka1_1234
# program ini menggunakan fungsi input()


print("\n========================================")
print("3. Operator Bitwise")
print("==========================================")

angka1_2031 = int(input("Masukkan angka bitwise-1: "))
angka2_2031 = int(input("masukkan angka bitwise-2: "))

print("\n angka dalam bentuk desimal dan biner")
print("angka1 =", angka1_2031, "| biner = ", bin(angka1_2031))
print("angka2 =", angka2_2031, "| biner = ", bin(angka2_2031))

# Bitwise AND
hasil_2031 = angka1_2031 & angka2_2031
print("\n Bitwise AND (&)")
print(angka1_2031, "&", angka2_2031, "=", hasil_2031)
print("Biner Hasil =", bin (hasil_2031))
print("Biner Hasil (8 bit) =", format(hasil_2031, "08b"))

# Bitwise OR
hasil_2031 = angka1_2031 | angka2_2031
print("\n Bitwise OR (|)")
print(angka1_2031, "|", angka2_2031, "=", hasil_2031)
print("Biner Hasil =", bin(hasil_2031))
print("Biner Hasil (8 bit) =", format(hasil_2031, "08b"))

# Bitwise XOR
hasil_2031 = angka1_2031 ^ angka2_2031
print("\n Bitwise XOR (^)")
print(angka1_2031, "^", angka2_2031, "=", hasil_2031)
print("Biner Hasil", bin(hasil_2031))
print("Biner Hasil (8 bit) =", format(hasil_2031, "08b"))

# Bitwise NOT
hasil_2031 = ~angka1_2031
print("\n Bitwise NOT (~)")
print("~", angka1_2031, "=", hasil_2031)
print("Biner Hasil", bin(hasil_2031))
print("Biner Hasil (8 bit) =", format(hasil_2031, "08b"))

# Bitwise geser kiri
jumlah_geser = int(input("\n masukkan jumlah pergeseran bit: "))

hasil_2031 = angka1_2031 << jumlah_geser
print("\n Bitwise geser kiri (<<)")
print(angka1_2031, "<<", jumlah_geser, "=", hasil_2031)
print("Biner Hasil", bin(hasil_2031))
print("Biner Hasil (8 bit) =", format(hasil_2031, "08b"))

# Bitwise geser kanan
hasil_2031 = angka1_2031 >> jumlah_geser
print("\n Bitwise geser kanan (>>)")
print(angka1_2031, ">>", jumlah_geser, "=", hasil_2031)
print("Biner Hasil", bin(hasil_2031))
print("Biner Hasil (8 bit) =", format(hasil_2031, "08b"))
```

<div style="font-family:'Times New Roman', Times, serif; font-size: 18px;">
Penjelasan:

1. Bentuk Desimal dan Biner<br>
Angka pertama yang digunakan adalah 4, sedangkan angka kedua adalah 8.<br>
Dalam bentuk biner:<br>
4 = 100<br>
8 = 1000<br>
Bentuk biner ini digunakan karena operator bitwise bekerja pada setiap bit dari bilangan tersebut.<br>
2. Bitwise AND (&)<br>
Operasi AND membandingkan setiap bit dari kedua angka. Hasilnya bernilai 1 hanya jika kedua bit yang dibandingkan sama-sama 1.<br>
  0100<br>
& 1000<br>
  0000<br>
Sehingga:<br>
4 & 8 = 0<br>
Hasil dalam 8 bit adalah 00000000.<br>
3. Bitwise OR (|)<br>
Operasi OR menghasilkan nilai 1 apabila salah satu atau kedua bit bernilai 1.<br>
  0100<br>
| 1000<br>
  1100<br>
Biner 1100 memiliki nilai desimal 12.<br>
Sehingga:<br>
4 | 8 = 12<br>
Hasil dalam 8 bit adalah 00001100.<br>
4. Bitwise XOR (^)<br>
Operasi XOR menghasilkan nilai 1 apabila kedua bit berbeda.<br>
  0100<br>
^ 1000<br>
  1100<br>
Karena posisi bit 1 pada angka 4 dan angka 8 berbeda, hasilnya adalah 1100, yaitu 12 dalam desimal.<br>
Sehingga:<br>
4 ^ 8 = 12<br>
Hasil dalam 8 bit adalah 00001100.<br>
5. Bitwise NOT (~)<br>
Operator NOT digunakan untuk membalik bit. Pada Python, operasi NOT menggunakan representasi bilangan bertanda sehingga hasil dari ~4 adalah -5.<br>
Secara matematis, operasi NOT pada bilangan 4 dapat dipahami dengan rumus:<br>
~n = -(n + 1)<br>
Maka:<br>
~4 = -(4 + 1) = -5<br>
Oleh karena hasilnya negatif, tampilan biner dengan format "08b" juga akan menampilkan tanda negatif, sehingga outputnya adalah:<br>
-00000101<br>
6. Bitwise Geser Kiri (<<)<br>
Operator geser kiri digunakan untuk menggeser bit ke sebelah kiri sebanyak jumlah pergeseran yang ditentukan.<br>
Jumlah pergeseran adalah 3.<br>
4 = 00000100<br>
     ↓ geser kiri 3 bit<br>
   00100000<br>
Hasilnya adalah 32.<br>
Sehingga:<br>
4 << 3 = 32<br>
Setiap pergeseran ke kiri satu posisi secara umum setara dengan mengalikan bilangan dengan 2. Karena digeser 3 kali:<br>
4 × 2³ = 4 × 8 = 32<br>
Hasil biner 8 bit adalah 00100000.<br>
7. Bitwise Geser Kanan (>>)<br>
Operator geser kanan digunakan untuk menggeser bit ke sebelah kanan sebanyak 3 posisi.<br>
4 = 00000100<br>
     ↓ geser kanan 3 bit<br>
   00000000<br>
Karena angka 4 digeser ke kanan sebanyak 3 posisi, semua bit 1 akan keluar dari posisi yang digunakan.<br>
Sehingga:<br>
4 >> 3 = 0<br>
Hasil biner 8 bit adalah 00000000.<br>

### 2.2.4 Program Operator Keanggotaan_2611532031

```python
print("=========================")
print(" 1. OPERATOR KEANGGOTAAN")
print("=========================")

# input beberapa data yang dipisahkan dengan koma
input_data_2031 = input("masukkan beberapa angka, pisahkan dengan koma: ")

# mengubah input menjadi list integer
data_2031 = [int(angka.strip()) for angka in input_data_2031.split(",")]

nilai_dicari_2031 = int(input("masukkan angka yang ingin di cari: "))

#operator in 
hasil_2031 = nilai_dicari_2031 in data_2031 
print("\n Operator keanggotaan IN")
print(nilai_dicari_2031, "in", data_2031, "=", hasil_2031)

# operator not in 
hasil_2031 = nilai_dicari_2031 not in data_2031 
print("\n Operator keanggotaan NOT IN")
print(nilai_dicari_2031, "not in", data_2031, "=", hasil_2031)
```

### 2.2.5 Program Operator Identitas_2611532031
</div>

```python
print("\n======================")
print("2. OPERATOR IDENTITAS")
print("========================")

# membuat data list dari input pengguna
input_data_2031 = input("Masukkan beberapa angka, pisahkan dengan koma: ")
data_2031 = [int(angka.strip()) for angka in input_data_2031.split(",")]

# objek1 menggunakan list dari input pengguna 
objek1_2031 = data_2031

# objek2 merujuk pada objek yang sama dengan objek1
objek2_2031 = objek1_2031 

# objek3 memiliki isi sama, tetapi merupakan objek baru
objek3_2031 = data_2031.copy()

print("objek1_2031 =", objek1_2031)
print("objek2_2031 =", objek2_2031)
print("objek3_2031 =",objek3_2031)

# operator is
# operator is not
hasil_2031 = objek1_2031 is not objek3_2031
print("\n Operator identitas IS NOT")
print("objek1_2031 is not objek3_2031 =", hasil_2031)
# operator is not
hasil_2031 = objek1_2031 is objek3_2031
print("\n Operator identitas IS NOT")
print("objek1_2031 is not objek3_2031 =", hasil_2031)

# membandingkan identitas dan nilai 
print("\n membandingkan identitas dan nilai")
print("objek1_2031 is objek3_2031: ", objek1_2031 is objek3_2031)
print("objek1_2031 == objek3_2031: ", objek1_2031 == objek3_2031)
```

<div style="font-family:'Times New Roman', Times, serif; font-size: 18px;">
Penjelasan:

1. Data Objek<br>
Data yang digunakan adalah 8 dan 7, sehingga data_2031 memiliki nilai [8, 7]. Kemudian data tersebut digunakan untuk membuat objek1_2031, objek2_2031, dan objek3_2031.<br>
2. Objek 1 dan Objek 2<br>
objek2_2031 merujuk langsung pada objek1_2031. Artinya, kedua variabel tersebut menunjuk pada objek yang sama. Oleh karena itu, operator is menghasilkan True.<br>
3. Objek 3<br>
objek3_2031 dibuat menggunakan proses copy(). Walaupun memiliki isi yang sama dengan objek1_2031, objek tersebut merupakan objek baru. Oleh sebab itu, identitas keduanya berbeda.<br>
4. Operator is<br>
Perbandingan objek1_2031 is objek2_2031 menghasilkan True karena kedua variabel merujuk pada objek yang sama.<br>
5. Operator is not<br>
Pada program terdapat perbandingan identitas antara objek1_2031 dan objek3_2031. Karena keduanya merupakan objek yang berbeda, maka kondisi is not bernilai True jika ditulis sesuai operator tersebut.<br>
Namun, pada kode yang diberikan, baris:<br>
objek1_2031 is not objek3_2031<br>
seharusnya menghasilkan True, bukan False. Jadi output yang benar untuk bagian tersebut adalah:<br>

```python
print ("\n=== SISTEM TRANSAKSI TOKO ===")

nama_2031 = input("Masukkan Nama Pelanggan: ")
status_2031 = input("Masukkan Status Pelanggan (member/nonmember): ")
total_belanja_2031 = int(input("Masukkan Total Belanja : "))
jumlah_barang_2031 = int(input("Masukkan Jumlah Barang : "))
promo_2031 = input("Masukkan Kode Promo : ")

 #========================================
 #  ======DATA TRANSAKSI======
 # =======================================


print("\n=== DATA TRANSAKSI ===")
print("Nama Pelanggan                     : ",nama_2031)
print("Status Pelanggan (member/nonmember): ", status_2031)
print(f"Total Belanja                      :  Rp{total_belanja_2031}")
print("Jumlah Barang                      : ", jumlah_barang_2031)
print("Kode Promo                         : ", promo_2031)

#=========================================
#=========HASIL VALIDASI================
#=========================================

kode_promo_2031 = ["HIDUP IF", "HIDUP FTI", "JALAN SEHAT"]

syarat_total_2031 = total_belanja_2031 >=200000
syarat_jumlah_barang_2031 = jumlah_barang_2031 >=3
valid_status_2031 = status_2031 == "member"
valid_promo_2031 = promo_2031 in kode_promo_2031 


print("\n=== HASIL VALIDASI ===")
print(f"Belanja >= Rp200000                : {syarat_total_2031}") 
print(f"Jumlah Barang >= 3                 : {syarat_jumlah_barang_2031}")
print(f"Status Member                      : {valid_status_2031}")
print(f"Kode Promo Tersedia                : {valid_promo_2031}")
print(f"Mendapatkan Diskon                 : {syarat_jumlah_barang_2031 or syarat_total_2031}")
print(f"Mendapatkan Promo                  : {valid_promo_2031}")

#=========================================
#  =====HASIL PERHITUNGAN============
#=========================================

print("\n=== HASIL PERHITUNGAN ===")
print(f"Diskon                             : Rp{int(0.15 * total_belanja_2031)}")
print(f"Total Pembayaran                   : Rp{int(total_belanja_2031 - 0.15 * total_belanja_2031)}")
print(f"Rata-rata Harga Barang             : Rp{int((total_belanja_2031 - 0.15 * total_belanja_2031) // jumlah_barang_2031)}")

#=========================================
#=========HAK AKSES PELANGGAN===========
#=========================================
 
print("\n=== HAK AKSES PELANGGAN ===")
print("Kode Hak Akses                     : ...")
print(f"Member Access                      : {status_2031 == "member"}")
print(f"Promo Access                       : {promo_2031 in kode_promo_2031}")
print("Free Shipping Access               : ...")

#=========================================
#=========OPERASI BITWISE===============
#=========================================


print("\n=== OPERASI BITWISE ===")

print("\n=== Kode Status Transaksi ===")

#KODE STATUS
#0001 = member
#0010 = total belanja >= Rp 200000
#0100 = jumlah barang >= 3
#1000 = promo

#menggunakan OR (|)
kode_transaksi_2031 =  int(valid_status_2031) << 0 | int(syarat_total_2031) << 1 | int(syarat_jumlah_barang_2031) << 2 | int(valid_promo_2031) << 3
kode_referensi_2031 = int(valid_status_2031) << 0 | int(syarat_total_2031) << 1 | int(valid_promo_2031) << 3

print(f"{format(int(valid_status_2031) << 0, "04b")} | {format(int(syarat_total_2031) << 1, "04b")} | {format(int(syarat_jumlah_barang_2031) << 2, "04b")} | {format(int(valid_promo_2031) << 3, "04b")}")
print(f"Kode Biner   : {format(kode_transaksi_2031, "04b")}") 
print(f"Kode Desimal : {kode_transaksi_2031}")

#====================================
#========PEMERIKSAAN STATUS=========
#====================================

print("\n=== Pemeriksaan Status ===")


print("\nCek Member")

#menggunakan and (&)

print(f"{format(kode_transaksi_2031, "04b")} & {format(int(valid_status_2031) << 0, "04b")}")
print(f"Hasil Biner   : {format((kode_transaksi_2031) & int(valid_status_2031) << 0, "04b")}")
print(f"Hasil Desimal : {(kode_transaksi_2031) & int(valid_status_2031) << 0}") 

print("\nCek Promo")

print(f"{format(kode_transaksi_2031, "04b")} & {format(int(valid_promo_2031) << 3, "04b")}")
print(f"Hasil Biner   : {format((kode_transaksi_2031) & int(valid_promo_2031) << 3, "04b")}")
print(f"Hasil Desimal  : {(kode_transaksi_2031) & int(valid_promo_2031) << 3}")

#====================================
#======PERBANDINGAN STATUS==========
#============================================

print("\n=== Perbandingan Status ===")

 # menggunakan XOR (^)

print(f"Kode Transaksi : {format(kode_transaksi_2031, "04b")}")
print(f"Kode Referensi : {format(kode_referensi_2031, "04b")}")
print(f"{format(kode_transaksi_2031, "04b")} ^ {format(kode_referensi_2031, "04b")}")
print(f"Hasil Biner    : {format((kode_transaksi_2031) ^ (kode_referensi_2031))}")
print(f"Hasil Desimal  : {(kode_transaksi_2031) ^ (kode_referensi_2031)}")

#=====================================
#=======SHIFT GESER KIRI=============
#=====================================


print("\n=== Shift ===")

#bitwise kiri/ shift kiri (<<)

print(f"{format(kode_transaksi_2031,"04b")} << 1")
print(f"Hasil Biner   : {format((kode_transaksi_2031) << 1,"04b")}")
print(f"Hasil Desimal : {(kode_transaksi_2031) << 1}") 
print("=== SELESAI ===")
```

Pengurangan digunakan untuk mendapatkan total pembayaran setelah diskon. Sedangkan pembagian bulat digunakan untuk menghitung rata-rata harga barang berdasarkan total pembayaran dan jumlah barang. Operator tersebut membantu program melakukan perhitungan transaksi secara otomatis.

2. Perbedaan operator and, or, dan not<br>
Operator and digunakan ketika semua kondisi harus terpenuhi agar menghasilkan nilai True. Operator or digunakan ketika cukup salah satu kondisi terpenuhi untuk menghasilkan True. Sedangkan operator not digunakan untuk membalik hasil suatu kondisi, yaitu True menjadi False dan False menjadi True.<br>
Pada program yang dibuat, konsep operator logika dapat digunakan untuk menentukan apakah pelanggan memenuhi beberapa persyaratan sekaligus, misalnya jumlah belanja dan jumlah barang harus memenuhi batas tertentu.

3. Fungsi operator in dan not in pada pemeriksaan kode promo<br>
Operator in digunakan untuk memeriksa apakah kode promo yang dimasukkan pelanggan terdapat dalam daftar kode promo yang tersedia. Jika kode yang dimasukkan terdapat dalam daftar, maka hasil pemeriksaannya adalah True.<br>
Sedangkan not in digunakan untuk memeriksa apakah kode promo tidak terdapat dalam daftar. Jika kode tersebut tidak tersedia, maka hasilnya adalah True.<br>
Dalam program, operator ini digunakan untuk menentukan apakah pelanggan mendapatkan akses terhadap promo yang tersedia.

4. Perbedaan is dan == pada Python<br>
Operator == digunakan untuk membandingkan nilai dari dua data. Sedangkan operator is digunakan untuk memeriksa apakah dua objek merupakan objek yang sama.<br>
Dalam program transaksi, operator == digunakan untuk membandingkan status pelanggan, misalnya untuk mengetahui apakah status pelanggan adalah member<br>
Operator is tidak digunakan karena program hanya membutuhkan perbandingan nilai, bukan pemeriksaan objek yang sama.

5. Penggunaan operator bitwise untuk menentukan hak akses pelanggan<br>
Operator bitwise digunakan untuk mengolah kode hak akses dalam bentuk bilangan biner. Setiap posisi bit dapat mewakili status atau hak akses tertentu, seperti member dan promo.<br>
Operator OR (|) digunakan untuk menggabungkan beberapa kode status menjadi satu kode transaksi. Kemudian operator AND (&) digunakan untuk memeriksa apakah suatu hak akses tertentu aktif.<br>
Selain itu, program menggunakan XOR (^) untuk membandingkan kode transaksi dengan kode referensi. Operator shift kiri (<<) digunakan untuk menggeser posisi bit ke sebelah kiri.<br>
Dengan cara tersebut, hak akses pelanggan dapat direpresentasikan menggunakan kode biner.

6. Alur program dari input hingga menghasilkan total pembayaran dan status promo<br>
Alur program dimulai dengan pelanggan memasukkan nama, status pelanggan, total belanja, jumlah barang, dan kode promo.<br>
Setelah data dimasukkan, program menampilkan kembali data transaksi tersebut. Selanjutnya program melakukan validasi untuk mengetahui apakah total belanja telah mencapai batas minimal, jumlah barang memenuhi syarat, pelanggan merupakan member, dan kode promo tersedia.<br>
Setelah proses validasi selesai, program menghitung diskon dan total pembayaran. Program juga menghitung rata-rata harga barang berdasarkan total pembayaran dan jumlah barang.<br>
Selanjutnya program menentukan hak akses pelanggan, seperti akses member dan akses promo. Terakhir, program melakukan operasi bitwise terhadap kode status transaksi untuk menunjukkan proses pemeriksaan hak akses dalam bentuk bilangan biner.

7. Contoh kondisi True dan False<br>
Contoh kondisi True:<br>
Jika pelanggan memiliki total belanja Rp600.000, jika kondisi belanja minimal Rp200.000 akan menghasilkan True, karena Rp600.000 memenuhi batas minimal yang ditentukan.<br>
Contoh kondisi False:<br>
Jika pelanggan membeli 2 barang, maka kondisi jumlah barang minimal 3 akan menghasilkan False, karena jumlah barang tersebut belum memenuhi persyaratan.<br>
Contoh lainnya, jika pelanggan memasukkan kode promo yang tidak terdapat dalam daftar promo, maka pemeriksaan kode promo dengan operator in akan menghasilkan False.<br>
