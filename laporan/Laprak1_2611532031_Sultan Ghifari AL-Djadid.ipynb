## 2.1 Dasar Teori
Pengenalan tentang Python, VS Code, Github

1. Visual Studio Code
<br>(VS Code) adalah aplikasi code editor yang digunakan untuk menulis, mengedit, dan menjalankan program. VS Code mendukung berbagai bahasa pemrograman, termasuk Python. Aplikasi ini memiliki fitur seperti debugging, terminal, ekstensi, serta integrasi dengan Git dan GitHub, sehingga memudahkan proses pembuatan dan pengelolaan proyek.
2. Python 
<br>Python adalah bahasa pemrograman yang digunakan untuk memberikan perintah kepada komputer agar melakukan sesuatu.
Python memiliki cara penulisan yang sederhana dan mudah dipahami, sehingga cocok untuk orang yang baru belajar pemrograman. Dengan Python, kita bisa membuat berbagai program, seperti mengolah data, membuat aplikasi, menghitung sesuatu, dan mengotomatisasi pekerjaan. contoh print("Hello World")
3. GitHub 
<br>Github adalah platform yang digunakan untuk menyimpan, mengelola, dan membagikan kode program secara online. GitHub membantu programmer menyimpan proyek dengan aman dan mencatat setiap perubahan yang dilakukan pada kode. GitHub juga dapat digunakan untuk bekerja sama dalam sebuah proyek. Beberapa orang dapat mengerjakan proyek yang sama tanpa harus mengirim file secara manual satu per satu.
4. Repository 
<br>Repository adalah tempat untuk menyimpan dan mengelola file dalam sebuah project. Repository dapat menyimpan kode, dokumen, serta riwayat perubahan project.
5. Commit dan Push
<br>Commit adalah proses untuk menyimpan perubahan pada project ke dalam Git. Commit biasanya diberikan pesan agar kita tahu perubahan apa yang telah dilakukan. Push adalah proses untuk mengirim atau mengupload hasil commit dari komputer ke repository yang ada di GitHub.

### 2.2.2 Membuat Repository Github

setelah program dibuat dan dilakukan uji coba pada VS Code, proyek dapat dihubungkan dengan Github. Repository digunakan untuk menyimpan kode program secara daring.

Tahapan yang dilakukan yaitu:

1. Membuat Repository baru pada Github.
2. Membuka folder proyek menggunakan VS Code 
3. Membuka terminal pada VS Code
4. Menginisialisasi Git pada folder proyek
5. Menambahkan file ke dalam staging 
6. Membuat commit
7. Menghubungkan repository lokal dengan repository Github
8. Melakukan push untuk mengunggah kode ke Github.

contoh perintah yang digunakan adalah:

```python
git init
git add .
git commit -m "initial commit"
git branch -M main
git remote add origin https://github.com./username/nama-repository.git
git push -u origin main
```

Penjelasan:

1. git init
digunakan untuk memulai Git di dalam folder project.
2. git add .
digunakan untuk memasukkan semua file project ke Git.
3. git commit
digunakan untuk menyimpan perubahan yang sudah dibuat.
4. git branch
Untuk membuat main sebagai branch utama.
5. git remote add origin
digunakan untuk menghubungkan project dengan repository GitHub.
6. git push
digunakan untuk mengupload project ke GitHub.
