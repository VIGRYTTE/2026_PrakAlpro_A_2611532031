## 2.1 Dasar Teori
Pengenalan tentang tipe data dan variabel

1. Int (Integer)<br>
Int adalah tipe data yang digunakan untuk menyimpan bilangan bulat, baik bilangan positif, negatif, maupun nol. Tipe data ini biasanya digunakan dalam perhitungan yang tidak membutuhkan angka desimal.

2. Float<br>
Float adalah tipe data yang digunakan untuk menyimpan bilangan yang memiliki angka desimal. Tipe data ini banyak digunakan dalam perhitungan yang membutuhkan hasil yang lebih tepat.

3. Komplek (Complex)<br>
Complex adalah tipe data yang digunakan untuk menyimpan bilangan kompleks, yaitu bilangan yang memiliki bagian nyata dan bagian imajiner. Tipe data ini biasanya digunakan dalam perhitungan matematika atau bidang tertentu yang membutuhkan bilangan kompleks.

4. Boolean<br>
Boolean adalah tipe data yang hanya memiliki dua nilai, yaitu benar (True) atau salah (False). Tipe data ini biasanya digunakan untuk menentukan suatu kondisi dalam program.

5. Konstanta<br>
Konstanta adalah nilai yang tetap dan tidak seharusnya berubah selama program dijalankan. Konstanta digunakan untuk menyimpan nilai yang memang ingin dipertahankan agar tidak berubah.

6. String<br>
String adalah tipe data yang digunakan untuk menyimpan teks atau kumpulan karakter. String dapat berupa huruf, kata, kalimat, angka yang dianggap sebagai teks, maupun simbol.

### 2.2.2 Program "Numerik_2611532031"
</div>

```python
#create a variable with integer value.
a_2031=100
print("The type of variable having value", a_2031, " is" , type(a_2031))

#create a variable with float value.
b_2031=10.2345
print("The type of variable having value", b_2031, " is" , type(b_2031))

#create a variable with complex value.
c_2031=100+3j
print("The type of variable having value", c_2031, " is", type(c_2031))
```

### 2.2.3 Program "Boolean_2611532031"
</div>

```python
# Buat file dengan nama Boolean_2611532031
# Nama variabel ditambah 4 digit nim terakhir contoh: nilai_1234
# Deklarasi variabel dengan tipe data Boolean
is_lulus_2031 = True
is_cumlaude_2031 = True

# Menggunakan Boolean
nilai_2031 = 85
batas_lulus_2031 = 75

# Menantukan nilai Boolean dari kondisi
status_kelulusan_2031 = nilai_2031 >= batas_lulus_2031 # Hasilnya akan True

print("=== Check Kelulusan ===")
print("Nilai: ",nilai_2031)
print("Apakah lulus?: ",status_kelulusan_2031)
if is_lulus_2031 and is_cumlaude_2031:
    print("Selamat, Anda lulus dengan predikat Cum Laude!")
```

2. Deklarasi is_cumlaude_2031 = True sebagai boolean

5. Deklarasi batas_lulus_2031 = 75 sebagai integer
6. Deklarasi status_kelulusan_2031 = nilai_2031 >= batas_lulus_2031 sebagai boolean

### 2.2.4 Program Input_2611532031
</div>

```python
# Buat file dengan nama Input_2611532031.py
# Buat Program untuk menghitung luas persegi panjang
# Nama variabel ditambah 4 digit nimm terakhir contoh: nama_2031
# Program ini menggunakan fungsi input() untuk meminta pengguna
# memasukan nilai panjang dan lebar persegi panjang
# Nilai yang dimasukkan akan dikonversi menjadi tipe data integer
panjang_2031 = input('Masukkan nilai panjang: ')
lebar_2031 = input('Masukkan nilai lebar: ')
luas_2031 = int(panjang_2031) * int(lebar_2031)
print("Luas =", luas_2031)
```

### 2.2.5 Program Konstanta_2611532031
</div>

```python
# Buat file dengan nama Konstanta_2611532031
# Program ini menggunakan konstanta untuk menghitun luas lingkaran
# nama variabel ditambah 4 digit nim terakhir contoh: jari_2031
from typing import Final
PI: Final = 3.14
print("pi: %f" % PI)
jari_2031 = float(input('Masukkan jari-jari lingkaran: '))
luas_2031 = PI * jari_2031 * jari_2031
print("Luas lingkaran dengan jari-jari %.2f adalah %.2f" % (jari_2031, luas_2031))
```

<div style="font-family:'Times New Roman', Times, serif; font-size: 18px;">
Penjelasan:

1. Pemanggilan Final dari library typing from typing import Final
2. Deklarasi PI: Final = 3.14 sebagai konstanta
3. Cetak ("Pi: %f" % (PI))
4. Nilai jari_2031 = float(input("Masukkan nilai jari-jari: "))
5. Nilai luas_2031 = PI * jari_2031
6. Cetak ("Luas lingkaran dengan jari-jari %.2f adalah %.2f" % (jari_2031,luas_2031))
