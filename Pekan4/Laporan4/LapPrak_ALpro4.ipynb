## 2.1 Dasar Teori
Pengenalan tentang Statement Conditional


1. Pengertian Conditional Statement<br>
Conditional statement (pernyataan kondisional) adalah struktur kontrol dalam pemrograman yang digunakan untuk menjalankan suatu blok kode tertentu berdasarkan hasil evaluasi sebuah kondisi atau ekspresi logika (boolean). Kondisi tersebut hanya memiliki dua kemungkinan nilai, yaitu True (benar) atau False (salah). Berdasarkan nilai inilah program menentukan alur eksekusi selanjutnya.

2. Konsep Percabangan (Branching)
Struktur conditional sering disebut juga sebagai struktur percabangan, karena program tidak lagi berjalan secara linear (satu alur), melainkan bercabang menjadi beberapa kemungkinan jalur eksekusi, tergantung pada kondisi yang diperiksa. Secara umum terdapat tiga bentuk percabangan:<br>
Percabangan satu kondisi → hanya menggunakan if<br>
Percabangan dua kondisi → menggunakan if dan else<br>
Percabangan banyak kondisi → menggunakan if, elif, dan else

3. Teori Dasar Masing-Masing Pernyataan<br>
a. if<br>
Pernyataan if berfungsi mengevaluasi suatu ekspresi boolean. Apabila hasil evaluasinya bernilai True, maka blok kode yang berada di dalamnya akan dieksekusi. Sebaliknya, jika bernilai False, blok kode tersebut akan dilewati begitu saja.<br>
b. elif<br>
Pernyataan elif (kependekan dari else if) digunakan untuk memeriksa kondisi tambahan apabila kondisi pada if sebelumnya bernilai False. Dalam satu program, elif dapat digunakan lebih dari satu kali untuk menangani banyak kemungkinan kondisi secara berurutan, dan sifatnya eksklusif — artinya hanya satu blok saja yang akan dieksekusi meskipun terdapat banyak elif.<br>
c. else<br>
Pernyataan else tidak memiliki kondisi (ekspresi boolean) dan berfungsi sebagai blok "penutup" yang akan dijalankan apabila seluruh kondisi pada if dan elif di atasnya bernilai False.

4. Alur Eksekusi (Flow of Control)<br>
Secara teori, sistem akan mengevaluasi kondisi-kondisi tersebut dari atas ke bawah secara berurutan, dengan tahapan sebagai berikut:<br>
Kondisi pada if diperiksa terlebih dahulu.<br>
Jika bernilai False, sistem akan melanjutkan pemeriksaan ke elif (jika ada), satu per satu secara berurutan.<br>
Apabila seluruh kondisi pada if maupun elif bernilai False, maka blok else yang akan dijalankan.<br>
Begitu ditemukan satu kondisi yang bernilai True, blok kode terkait akan dieksekusi, dan seluruh percabangan lain yang tersisa tidak akan diperiksa lagi (diabaikan).<br>

5. Operator yang Umum Digunakan dalam Kondisi<br>
Kondisi yang diperiksa pada if maupun elif biasanya dibentuk dengan menggunakan dua jenis operator, yaitu:<br>
Operator perbandingan, seperti sama dengan, tidak sama dengan, lebih besar, lebih kecil, lebih besar atau sama dengan, dan lebih kecil atau sama dengan.<br>
Operator logika, seperti and (dan), or (atau), dan not (bukan/negasi), yang digunakan untuk menggabungkan atau membalik beberapa kondisi sekaligus
.
6. Sifat Penting Conditional Statement<br>
Bersifat eksklusif mutual, artinya hanya satu blok kode yang akan dijalankan dari keseluruhan struktur if-elif-else.<br>
Urutan pemeriksaan kondisi sangat berpengaruh terhadap hasil akhir program, sehingga penyusunan urutan kondisi perlu dilakukan secara cermat dan logis.<br>
Struktur ini menjadi dasar penting sebelum mempelajari konsep pemrograman lanjutan lainnya, seperti perulangan (looping) dan fungsi (function).

## 2.2 Kode Program

### 2.2.1 Program "IF1_2611532031"
</div>

```python
#Buat file dengan nama If1_2611532031.py
#Buat program untuk kondisional if
#Nama variabel ditambah 4 digit nim terakhir contoh: ipk_2031
#Program ini menggunakan fungsi input()

IPK_2031 = float(input("Input IPK Anda = "))

if IPK_2031 > 2.75:
    print("Anda Lulus Sangat Memuaskan dengan IPK " + str(IPK_2031))
    
print("Program Selesai")
```

<div style="font-family:'Times New Roman', Times, serif; font-size: 18px;">

Penjelasan:
1. Input Data<br>
Program meminta pengguna memasukkan nilai IPK melalui fungsi input().<br>
Karena hasil dari input() selalu berupa teks (string), nilai tersebut langsung diubah menjadi tipe data desimal (float) agar bisa dibandingkan secara matematis, bukan dibandingkan sebagai teks.<br><br>
2. Struktur Percabangan (if tunggal)<br>
Program menggunakan satu kondisi if saja, tanpa pasangan else.<br>
Kondisi yang dicek adalah apakah IPK yang diinput lebih besar dari 2.75.<br>
Jika IPK yang dimasukkan lebih besar dari 2.75, maka kondisi bernilai benar, sehingga program mencetak pesan kelulusan yang menyatakan pengguna lulus dengan predikat sangat memuaskan beserta nilai IPK-nya.<br>
Jika IPK yang dimasukkan kurang dari atau sama dengan 2.75, maka kondisi bernilai salah, sehingga pesan kelulusan tersebut tidak dicetak sama sekali.<br>
Karena tidak ada else sebagai penangkap kondisi sebaliknya, tidak ada pesan alternatif apa pun yang muncul untuk kasus ini.<br><br>
Pada bagian pencetakan pesan, nilai IPK yang bertipe angka digabungkan dengan teks menggunakan operator penggabungan string.<br>
Karena angka tidak bisa langsung digabung dengan teks di Python, nilai IPK tersebut terlebih dahulu diubah kembali menjadi bentuk teks sebelum digabungkan — proses ini disebut type casting (konversi tipe data).<br><br>
3. Penutup Program<br>
Di baris terakhir, program mencetak pesan "Program Selesai".<br>
Pesan ini selalu muncul, apa pun hasil dari pengecekan IPK sebelumnya — baik IPK-nya memenuhi syarat maupun tidak — karena baris ini berada di luar struktur if, sehingga dieksekusi secara terpisah setelah blok kondisi selesai diproses.<br><br>
Contoh Skenario:<br>
Jika pengguna memasukkan IPK 3.5 (lebih besar dari 2.75), maka akan muncul pesan kelulusan sangat memuaskan, diikuti pesan "Program Selesai".<br>
Jika pengguna memasukkan IPK 2.5 (lebih kecil dari 2.75), maka pesan kelulusan tidak akan muncul, dan program langsung menampilkan "Program Selesai" saja.<br><br>

### 2.2.2 Program "IF2_2611532031"
</div>

```python
#Buat file dengan nama If2_2611532031.py
#Buat program untuk kondisional if
#Nama variabel ditambah 4 digit nim terakhir contoh: ipk_2031
#Program ini menggunakan fungsi input()

IPK_2031 = float(input("Input IPK_2031 Anda = "))

if IPK_2031 > 2.75:
    print("Anda Lulus Sangat Memuaskan dengan IPK " + str(IPK_2031))
else:
    print("Anda Tidak Lulus")
print("Program Selesai")
```

<div style="font-family:'Times New Roman', Times, serif; font-size: 18px;">
Penjelasan:

1. Input Data<br>
Program meminta pengguna memasukkan nilai IPK melalui fungsi input().<br>
Karena hasil dari input() selalu berupa teks (string), nilai tersebut langsung diubah menjadi tipe data desimal (float) agar bisa dibandingkan secara matematis, bukan dibandingkan sebagai teks.<br><br>
2. Struktur Percabangan (if-else)<br>
Program menggunakan struktur if-else, yaitu dua kemungkinan hasil yang saling melengkapi (jika kondisi benar dijalankan satu blok, jika salah dijalankan blok yang lain).<br>
Kondisi yang dicek adalah apakah IPK yang diinput lebih besar dari 2.75.<br>
Jika IPK yang dimasukkan lebih besar dari 2.75, maka kondisi if bernilai benar, sehingga program mencetak pesan kelulusan yang menyatakan pengguna lulus dengan predikat sangat memuaskan beserta nilai IPK-nya.<br>
Jika IPK yang dimasukkan kurang dari atau sama dengan 2.75, maka kondisi if bernilai salah, sehingga program otomatis menjalankan blok else, yaitu mencetak pesan "Anda Tidak Lulus".<br>
Berbeda dengan struktur if tunggal, pada if-else ini selalu ada satu pesan yang tercetak — tidak ada kemungkinan program tidak mencetak apa pun, karena setiap kondisi sudah memiliki pasangan penanganannya masing-masing (benar → blok if, salah → blok else).<br><br>
Pada bagian pencetakan pesan kelulusan, nilai IPK yang bertipe angka digabungkan dengan teks menggunakan operator penggabungan string.<br>
Karena angka tidak bisa langsung digabung dengan teks di Python, nilai IPK tersebut terlebih dahulu diubah kembali menjadi bentuk teks sebelum digabungkan — proses ini disebut type casting (konversi tipe data).<br>
Pesan pada blok else ("Anda Tidak Lulus") tidak menyertakan nilai IPK, sehingga tidak memerlukan proses penggabungan atau konversi tipe data apa pun.<br><br>
Contoh Skenario:<br>
Jika pengguna memasukkan IPK 3.5 (lebih besar dari 2.75), maka kondisi if terpenuhi dan akan muncul pesan kelulusan sangat memuaskan beserta nilai IPK-nya.<br>
Jika pengguna memasukkan IPK 2.5 (lebih kecil dari 2.75), maka kondisi if tidak terpenuhi, sehingga program menjalankan blok else dan mencetak pesan "Anda Tidak Lulus".<br><br>
Poin Penting Struktur if-else:<br>
Ciri khas if-else adalah ia selalu menangani dua kemungkinan sekaligus — kondisi benar dan kondisi salah — sehingga program pasti akan mencetak salah satu dari dua pesan yang tersedia, tergantung hasil evaluasi kondisinya.<br>
Ini berbeda dengan if tunggal yang hanya menangani satu kemungkinan (benar) dan membiarkan kondisi sebaliknya tanpa penanganan apa pun.

### 2.2.3 Program "MULTI_IF1_2611532031
</div>

```python
#Buat file dengan nama Multi_If1_2611532031.py
#Buat program untuk kondisional if
#Nama variabel ditambah 4 digit nim terakhir contoh: ipk_2031
#Program ini menggunakan fungsi input()

Umur_2031 = int(input("Input Umur Anda: "))
Sim_2031 = input("Apakah Anda Sudah Punya SIM C (y/t): ")[0].lower()

if Umur_2031 >= 17 and Sim_2031 == 'y':
    print("Anda Sudah Dewasa dan Boleh Bawa Motor")

elif Umur_2031 >= 17 and Sim_2031 != 'y':
    print("Anda Sudah Dewasa tetapi Tidak Boleh Bawa Motor")

elif Umur_2031 <= 17 and Sim_2031 == 'y':
    print("Anda Belum Cukup Umur untuk Punya SIM")


else :
    print("Anda Belum Cukup Umur dan Tidak Boleh Bawa Motor")
print("Program Selesai")
```

<div style="font-family:'Times New Roman', Times, serif; font-size: 18px;">
Penjelasan:

1. Input Data<br>
Program meminta pengguna memasukkan umur melalui fungsi input(), lalu hasilnya langsung diubah menjadi tipe data bilangan bulat (integer) agar bisa dibandingkan secara matematis.<br>
Program juga meminta status kepemilikan SIM C dengan jawaban "y" (ya) atau "t" (tidak).<br>
Dari jawaban tersebut, hanya karakter pertamanya saja yang diambil menggunakan pengindeksan string [0], lalu langsung disimpan ke dalam variabel — sehingga meskipun pengguna mengetik "ya" atau "yes", yang tersimpan tetap huruf pertamanya saja, yaitu "y".<br><br>
2. Struktur Percabangan (Multi-if Terpisah dengan Operator Logika)<br>
Program ini menggunakan empat blok if yang berdiri sendiri-sendiri (bukan if-elif-else berantai), sehingga setiap blok dievaluasi secara independen satu per satu tanpa saling menggugurkan.<br>
Setiap kondisi menggabungkan dua syarat sekaligus menggunakan operator logika and, yaitu kombinasi antara umur dan status kepemilikan SIM C.<br><br>
Blok pertama mengecek apakah umur lebih besar sama dengan 17 dan sudah punya SIM C.<br>
Jika kedua syarat tersebut terpenuhi bersamaan, maka dicetak pesan bahwa pengguna sudah dewasa dan boleh membawa motor.<br><br>
Blok kedua mengecek apakah umur lebih besar sama dengan 17 tetapi belum/tidak punya SIM C (menggunakan operator tidak sama dengan).<br>
Jika kedua syarat tersebut terpenuhi bersamaan, maka dicetak pesan bahwa pengguna sudah dewasa tetapi tidak boleh membawa motor karena belum memiliki SIM.<br><br>
Blok ketiga mengecek apakah umur kurang dari 17 tetapi sudah punya SIM C.<br>
Jika kedua syarat tersebut terpenuhi bersamaan, maka dicetak pesan bahwa pengguna belum cukup umur namun sudah memiliki SIM — kasus ini menandakan kejanggalan data karena secara aturan seharusnya SIM baru bisa dimiliki di atas umur tertentu.<br><br>
Blok keempat mengecek apakah umur kurang dari 17 dan juga belum/tidak punya SIM C.<br>
Jika kedua syarat tersebut terpenuhi bersamaan, maka dicetak pesan bahwa pengguna belum cukup umur untuk membawa motor.<br><br>
3. Karakteristik Penting Multi-if Terpisah<br>
Karena keempat blok ini ditulis sebagai if yang terpisah (bukan elif), Python akan tetap mengevaluasi seluruh empat kondisi tersebut satu per satu dari atas ke bawah, meskipun salah satu kondisi di atasnya sudah terpenuhi.<br>
Namun, karena keempat kondisi tersebut saling eksklusif (tidak mungkin dua kondisi sekaligus bernilai benar untuk kombinasi umur dan SIM yang sama), maka pada praktiknya hanya satu pesan saja yang akan tercetak untuk setiap kali program dijalankan.<br>
Berbeda dengan struktur if-elif-else yang otomatis berhenti begitu satu kondisi terpenuhi, struktur multi-if ini tetap memeriksa semua kondisi hingga blok terakhir, sehingga cocok digunakan saat beberapa hasil berpotensi ingin diakumulasikan (meski dalam kasus ini hasilnya tetap saling eksklusif).<br><br>

### 2.2.4 Program "MULTI_IF2_2611532031
</div>

```python
#Buat file dengan nama Multi_If1_2611532031.py
#Buat program untuk kondisional if
#Nama variabel ditambah 4 digit nim terakhir contoh: ipk_2031
#Program ini menggunakan fungsi input()
#Program Menghitung Diskon Belanja

#Input dari user
Total_Belanja_2031 = float(input("Masukkan Total Belanja (Rp): "))

#Input status member (mengecek apakah user mengetik 'y' atau 'ya')
Input_Member_2031 = input("Apakah Anda Member? (y/ya): ").strip().lower()
Is_Member_2031 = Input_Member_2031 in ["y", "ya"]

#Input status kode promo (mengecek apakah user mengetik 'y' atau 'ya')
Input_Promo_2031 = input("Apakah Kode Promo Valid? (y/ya): ").strip().lower()
Kode_Promo_2031 = Input_Promo_2031 in ["y", "ya"]

Total_Diskon_Persen = 0

#Multi-IF terpisah: Setiap kondisi diperiksa secara Independen
#Diskon Bisa Ditumpuk (akumulasi) jika memenuhi beberapa syarat sekaligus

if Total_Belanja_2031 > 1000000: 
    Total_Diskon_Persen += 10 # Diskon Belanja Besar

if Is_Member_2031:
    Total_Diskon_Persen += 5 # Diskon Member

if Kode_Promo_2031:
    Total_Diskon_Persen += 15 # Diskon Voucher

#Menghitung nominal diskon dan total bayar
Nominal_Diskon_2031 = Total_Belanja_2031 * (Total_Diskon_Persen / 100)
Total_Bayar_2031 = Total_Belanja_2031 - Nominal_Diskon_2031

#Output hasil 
print("\n--- Rincian Pembayaran ---")
print(f"Total Diskon : {Total_Diskon_Persen}% (Rp {Nominal_Diskon_2031:,.0F})")
print(f"Total Bayar : Rp {Total_Bayar_2031:,.0F}")

print(f"Total Diskon yang Anda Dapatkan: {Total_Diskon_Persen}%")
#Output Total Diskon yang Anda Dapatkan: 30% jika belanja > 1 juta, member, dan kode promo valid
```

<div style="font-family:'Times New Roman', Times, serif; font-size: 18px;">
Penjelasan:

1. Input Data<br>
Program meminta pengguna memasukkan total belanja melalui fungsi input(), lalu hasilnya langsung diubah menjadi tipe data desimal (float) agar bisa digunakan dalam perhitungan matematis.<br>
Program juga meminta status keanggotaan member dengan jawaban "y" atau "t", di mana teks yang diinput dirapikan terlebih dahulu menggunakan strip() (menghapus spasi berlebih) dan lower() (mengubah menjadi huruf kecil), sehingga input seperti " Y " atau "YA" tetap bisa terbaca dengan benar.<br>
Hasil jawaban tersebut kemudian dicek apakah termasuk dalam daftar ["y", "ya"], dan hasil pengecekan ini (bernilai benar atau salah) langsung disimpan ke dalam sebuah variabel bertipe boolean.<br>
Proses yang sama juga dilakukan untuk input status kode promo, sehingga didapat variabel boolean lain yang menandakan apakah kode promo yang dimasukkan valid atau tidak.<br><br>
2. Struktur Percabangan (Multi-if Terpisah untuk Akumulasi Diskon)<br>
Program ini menggunakan tiga blok if yang berdiri sendiri-sendiri (bukan elif), sehingga setiap kondisi dievaluasi secara independen dan diskonnya bisa ditumpuk atau diakumulasikan, bukan saling menggugurkan satu sama lain.<br><br>
Blok pertama mengecek apakah total belanja lebih besar dari Rp 1.000.000.<br>
Jika kondisi ini terpenuhi, maka persentase diskon ditambahkan sebesar 10% sebagai diskon belanja besar.<br><br>
Blok kedua mengecek apakah pengguna berstatus member (berdasarkan variabel boolean yang telah dihasilkan sebelumnya).<br>
Jika kondisi ini terpenuhi, maka persentase diskon ditambahkan sebesar 5% sebagai diskon member.<br><br>
Blok ketiga mengecek apakah kode promo yang dimasukkan valid (berdasarkan variabel boolean kode promo).<br>
Jika kondisi ini terpenuhi, maka persentase diskon ditambahkan sebesar 15% sebagai diskon promo.<br><br>
Karena ketiga blok tersebut ditulis sebagai if yang terpisah, seluruh kondisi akan tetap dicek satu per satu meskipun kondisi sebelumnya sudah terpenuhi, sehingga jika pengguna memenuhi ketiga syarat sekaligus (belanja di atas Rp 1.000.000, berstatus member, dan memiliki kode promo valid), maka ketiga persentase diskon tersebut akan dijumlahkan seluruhnya menjadi satu total diskon akhir.<br><br>
3. Perhitungan Nominal Diskon dan Total Bayar<br>
Setelah total persentase diskon terkumpul dari ketiga blok if di atas, nominal diskon dihitung dengan mengalikan total belanja dengan persentase diskon tersebut (dibagi 100 untuk mengubahnya dari bentuk persen ke bentuk desimal).<br>

### 2.2.7 Program MACTH_CASE1_2611532031
</div>

```python
#Buat file dengan nama Match_Case1_2611532031.py
#Buat program untuk match case
#Nama variabel ditambah 4 digit nim terakhir contoh: ipk_2031
#Program ini menggunakan fungsi input()
#Program konversi angka menjadi nama bulan

Bulan_2031 = int(input("Masukkan Angka Bulan (1 - 12): " ))

match Bulan_2031:
    case 1:
        print("Januari")
    case 2:
        print("Februari")
    case 3:
        print("Maret")
    case 4:
        print("April")
    case 5:
        print("Mei")
    case 6:
        print("Juni")
    case 7:
        print("Juli")
    case 8:
        print("Agustus")
    case 9:
        print("September")
    case 10:
        print("November")
    case 11:
        print("Oktober")
    case 12:
        print("Desember")
    case _:
        print("Angka Tidak Valid")
```

<div style="font-family:'Times New Roman', Times, serif; font-size: 18px;">
Penjelasan:

1. Input Data<br>
Program meminta pengguna memasukkan angka bulan (1 sampai 12) melalui fungsi input(), lalu hasilnya langsung diubah menjadi tipe data bilangan bulat (integer) agar bisa dicocokkan dengan nilai-nilai angka pada struktur match-case.<br><br>
2. Struktur Percabangan (match-case)<br>
Program ini menggunakan struktur match-case, yaitu bentuk percabangan yang mencocokkan satu nilai variabel dengan beberapa kemungkinan pola (case) yang telah ditentukan, mirip seperti if-elif-else namun dengan penulisan yang lebih ringkas dan rapi untuk kasus pencocokan nilai tunggal.<br>
Nilai yang diinput oleh pengguna dicocokkan satu per satu terhadap dua belas case yang tersedia, masing-masing mewakili angka 1 sampai 12.<br><br>
Jika angka yang diinput cocok dengan salah satu case, misalnya case 1, maka program akan mencetak nama bulan yang sesuai, yaitu "Januari".<br>
Begitu pula seterusnya, case 2 akan mencetak "Februari", case 3 mencetak "Maret", hingga case 12 yang mencetak "Desember" — setiap case mewakili satu bulan yang berurutan sesuai kalender.<br><br>
Begitu satu case yang cocok ditemukan, Python akan langsung menjalankan blok print() di dalamnya dan tidak lagi memeriksa case-case lain di bawahnya, sehingga hanya satu nama bulan saja yang akan tercetak untuk setiap kali program dijalankan.<br><br>
Case terakhir, yaitu case _, berfungsi sebagai case default atau penangkap semua kemungkinan nilai yang tidak cocok dengan case manapun di atasnya (angka 1 sampai 12).<br>
Jika pengguna memasukkan angka di luar rentang 1-12, misalnya 0, 13, atau angka negatif, maka program akan otomatis masuk ke case ini dan mencetak pesan "Angka tidak valid" sebagai penanda bahwa input yang dimasukkan salah.<br><br>
3. Karakteristik Penting Struktur match-case<br>
Struktur match-case sangat cocok digunakan ketika kita perlu mencocokkan satu variabel dengan banyak kemungkinan nilai spesifik (seperti pada kasus ini, dua belas kemungkinan angka bulan), karena penulisannya lebih terstruktur dan mudah dibaca dibandingkan menuliskan dua belas kondisi if-elif secara berurutan.<br>
Case _ yang berperan sebagai default sangat penting untuk selalu disertakan di bagian akhir, agar program tetap bisa menangani input yang tidak terduga atau di luar rentang yang valid, sehingga program tidak berhenti secara tiba-tiba (error) maupun tidak mencetak apa pun sama sekali.<br><br>
Contoh Skenario:<br>
Jika pengguna memasukkan angka 6, maka program akan mencocokkan dengan case 6 dan mencetak "Juni".<br>
Jika pengguna memasukkan angka 15 (di luar rentang 1-12), maka tidak ada satu case pun yang cocok, sehingga program masuk ke case _ dan mencetak pesan "Angka tidak valid".<br><br>

<h1 align="center"><b>BAB III<br>KESIMPULAN</b></h1>
<p style="text-indent: 40px; text-align: justify; "> Conditional Statement merupakan salah satu struktur kontrol paling fundamental dalam pemrograman, yang memungkinkan sebuah program untuk "mengambil keputusan" dan menentukan blok kode mana yang akan dijalankan berdasarkan hasil evaluasi kondisi logika (True atau False), bukan berjalan secara linear dari atas ke bawah tanpa arah alternatif. Struktur ini dibangun dari tiga elemen utama yang saling melengkapi, yaitu if sebagai pemeriksa kondisi awal, elif sebagai penambah kemungkinan kondisi lain jika kondisi sebelumnya tidak terpenuhi, dan else sebagai blok penutup yang menampung seluruh kemungkinan yang tersisa. Ketiganya dapat dikombinasikan sesuai kebutuhan, mulai dari percabangan sederhana satu kondisi (if saja), dua kemungkinan (if-else), hingga banyak kemungkinan sekaligus (if-elif-else). Dalam proses eksekusinya, sistem selalu memeriksa kondisi secara berurutan dari atas ke bawah, dan begitu ditemukan satu kondisi yang bernilai True, blok kode terkait langsung dijalankan sementara seluruh kondisi lain yang tersisa akan diabaikan (bersifat eksklusif mutual). Karena sifat inilah, urutan penyusunan kondisi menjadi sangat penting dan harus disusun secara logis dan cermat, sebab urutan yang keliru dapat menghasilkan alur eksekusi program yang tidak sesuai dengan yang diharapkan.Untuk membentuk kondisi yang akan dievaluasi, conditional statement memanfaatkan dua kelompok operator, yaitu operator perbandingan (untuk membandingkan nilai) dan operator logika (untuk menggabungkan atau membalik beberapa kondisi sekaligus), sehingga memungkinkan pembuatan aturan pengambilan keputusan yang fleksibel dan kompleks sesuai kebutuhan program. Secara keseluruhan, pemahaman yang kuat terhadap conditional statement menjadi fondasi penting sebelum mempelajari konsep-konsep pemrograman yang lebih lanjut, seperti perulangan (looping) dan fungsi (function), karena hampir seluruh logika pengambilan keputusan dalam program — baik yang sederhana maupun kompleks — pada dasarnya dibangun di atas prinsip percabangan (branching) yang telah dijelaskan dalam teori ini.
</p>
